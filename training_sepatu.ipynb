## BUAT DAPET AKURASI, F1 SCORE, RECALL

```python
# Muat model terbaik
model = YOLO('/content/drive/MyDrive/YOLO_Training_Shoe/experiment-9/weights/best.pt')

# Jalankan validasi pada test set
metrics = model.val(data=f"/content/shoe_material-5/data.yaml", split='test')

precision = metrics.seg.mp
recall = metrics.seg.mr

f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"Presisi (Mean Precision) : {precision:.4f}")
print(f"Recall (Mean Recall)       : {recall:.4f}")
print(f"F1-Score                   : {f1_score:.4f}")
print(f"mAP@50                     : {metrics.seg.map50:.4f}")
```
