```python
# Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
# 1. Load the dataset
df = pd.read_csv('voice.csv')

# Separate features (X) and target labels (y)
X = df.drop('label', axis=1)
y = df['label']

# Scale features for optimal SVM performance
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define configurations for split ratios and kernels
split_ratios = {'70:30': 0.30, '80:20': 0.20}
kernels = {'Linear': 'linear', 'Polynomial': 'poly', 'RBF': 'rbf'}

performance_results = []

# Loop through split ratios and kernels
for ratio_name, test_size in split_ratios.items():
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y, test_size=test_size, random_state=42
    )
    
    for kernel_name, kernel_type in kernels.items():
        # Initialize and train the SVM model
        svm_model = SVC(kernel=kernel_type, random_state=42)
        svm_model.fit(X_train, y_train)
        
        # Predict and calculate accuracy
        y_pred = svm_model.predict(X_test)
        acc = accuracy_score(y_test, y_pred) * 100  # as a percentage
        
        performance_results.append({
            'Split Ratio (Train:Test)': ratio_name,
            'Kernel': kernel_name,
            'Accuracy (%)': round(acc, 2)
        })

# Create and display the performance summary table
performance_df = pd.DataFrame(performance_results)
print(performance_df.to_markdown(index=False))
```
